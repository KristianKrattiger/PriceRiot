# PriceRiot Basic Analysis

This notebook demonstrates a simple end-to-end analytics flow on top of the PriceRiot simulation engine:

1. Run a headless simulation via the `simulation` Python extension.
2. Use the `python.analytics` helpers to collect transactions and customers into pandas DataFrames.
3. Compute a few basic summary statistics suitable for portfolio/demo use.

> Prerequisites: build the C++ simulation module first (see `README.md`), ensuring `PYTHONPATH` includes the `build/` directory.

```python
import os
import pandas as pd

import analytics  # if project root is on PYTHONPATH

# Run a short simulation for demo purposes
result = analytics.run_simulation(
    store_path="store.yaml",
    duration_seconds=300.0,
    spawn_interval=5.0,
    mission_probability=1,
    seed=42,
)

transactions = result.transactions
customers = result.customers

transactions.head(), customers.head()
```

```python
# Example: basic basket composition stats

basket_sizes = transactions.groupby("transaction_id")["quantity"].sum()
print("Average basket size:", basket_sizes.mean())
print("Top 10 SKUs by frequency:")
print(
    transactions.groupby("item_name")["quantity"].sum().sort_values(ascending=False).head(10)
)
```
